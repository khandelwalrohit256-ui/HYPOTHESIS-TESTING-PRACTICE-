```python
import numpy as np
import seaborn as sns
from scipy.stats import chi2_contingency
import pandas as pd
from scipy import stats
```

```python
df = sns.load_dataset('titanic')
```

```python
df.head(5)
```

```python
contingency_table = pd.crosstab(df['sex'],df['survived'])
contingency_table
```

```python
chi2 ,p_value,dof,expected = chi2_contingency(contingency_table)
```

```python
chi2
```

```python
expected
```

```python
dof
```

```python
p_value
```

```python
alpha = 0.05
```

```python
if p_value < alpha:
    print('we reject the null hypothesis and there is a significant relationship between gender and surival')
else:
    print('there is a no connection')
```
