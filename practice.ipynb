```python
#degrees of freedom
k = len(n)
N = np.sum(n)
```

```python
df_between = k-1
df_within = N-k
```

alternative hypothesis = the actual defect rate is not 5%.
Z-test for a Single Proportion isme ye test lgega kyoki yah ek proportion(pratishat/anuopat) ki samasya h .


```python
p_0 = 0.05 #  Claimed/Population proportion
n = 200 # sample size
alpha = 0.05 
defects = 16

p_hat =  defects / n #sample proportion
se = np.sqrt((p_0 * (1 - p_0)) / n)
z_stats = (p_hat - p_0) / se
z_stats
```

```python
p_value =  2*(norm.sf(abs(z_stats)))
p_value
```

```python
if p_value < alpha:
    print('reject null hypothesis')
else:
    print('fail to reject null hypothesis')
```
