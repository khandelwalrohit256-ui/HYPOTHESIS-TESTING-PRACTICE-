# 1st que 

```python
import numpy as np
from scipy import stats 
from scipy.stats import norm
```

```python
sample_mean = 504
population_mean = 500
population_std = 12
n = 36
alpha = 0.05
```

```python
z_score = (sample_mean - population_mean) / (population_std/np.sqrt(n))
```

```python
z_score 
```

```python
p_value = 2 * (1-norm.cdf(abs(z_score)))
```

```python
p_value
```

```python
if p_value < alpha:
    print('reject the null hypothesis')
else :
    print('accept the null hypothesis')
```

# Paired T-Test 
Jab same people k two related measurements compare krne hote h tab paired t test use krte h 

```python
# for example ham kuchh students k coaching jane se phle k marks or jane k bad k marks ko compare krte h
before = np.array([60,65,70,55,75])
after = np.array([70,68,78,60,80])

difference = after - before 
n = len(difference)
print(difference)
```

```python
mean_diff = np.mean(difference)
mean_diff
```

```python
std_diff = np.std(difference, ddof=1) # ddof = 1 use kiya h kyuki hame sample ka std chahiye kyuki sample k std me n-1 se devide hota h na ki n se or ye by default set hota h numpy me population k liye ddof = 0 
print(std_diff)
```

t_score = mean_diff/(std_diff/squre root of n)  here n is lenth of difference

```python
t_score = mean_diff / (std_diff / np.sqrt(n))
t_score
```

```python
p_value = 2 * stats.t.sf(abs(t_score), df=n-1)
p_value
```

```python
if p_value < alpha:
    print('reject_null hypothesis')
else:
    print('fail to reject null hypothesis')
```
